Q1: What are the key differences between SQL and NoSQL databases?
Ans:-
- SQL: Relational, structured schema, uses tables and rows, supports ACID transactions.
- NoSQL: Non‑relational, flexible schema, supports documents, key‑value, column, or graph storage, designed for scalability and high performance.

-------------------------------------------------------------------------------------------------------------------------------------------------------
Q2: What makes MongoDB a good choice for modern applications?
Ans:-
- Schema flexibility with JSON‑like documents.
- Horizontal scalability via sharding.
- High availability through replication.
- Rich query language and aggregation framework.

-------------------------------------------------------------------------------------------------------------------------------------------------------
Q3: Explain the concept of collections in MongoDB.
Ans:-
- A collection is a group of MongoDB documents, similar to a table in SQL.
- Collections do not enforce a strict schema, allowing flexible document structures.

-------------------------------------------------------------------------------------------------------------------------------------------------------
Q4: How does MongoDB ensure high availability using replication?
Ans:-
- MongoDB uses replica sets: one primary node and multiple secondary nodes.
- If the primary fails, an automatic election promotes a secondary to primary, ensuring uptime.

-------------------------------------------------------------------------------------------------------------------------------------------------------
Q5: What are the main benefits of MongoDB Atlas?
Ans:-
- Fully managed cloud service.
- Automated backups, scaling, and monitoring.
- Global distribution and multi‑cloud support.
- Built‑in security and compliance features.

-------------------------------------------------------------------------------------------------------------------------------------------------------
Q6: What is the role of indexes in MongoDB, and how do they improve performance?
Ans:-
- Indexes improve query performance by reducing the number of documents scanned.
- Example: An index on email speeds up searches for users by email.

-------------------------------------------------------------------------------------------------------------------------------------------------------
Q7: Describe the stages of the MongoDB aggregation pipeline.
Ans:-
- $match: Filter documents.
- $group: Aggregate values.
- $project: Reshape documents.
- $sort: Order results.
- $lookup: Join collections.
- $limit / $skip: Pagination.

-------------------------------------------------------------------------------------------------------------------------------------------------------
Q8: What is sharding in MongoDB? How does it differ from replication?
Ans:-
- Sharding: Splits data across nodes for scalability.
- Replication: Duplicates data across nodes for availability.
- Replication = redundancy; Sharding = distribution.

-------------------------------------------------------------------------------------------------------------------------------------------------------
Q9: What is PyMongo, and why is it used?
Ans:-
- PyMongo: Official Python driver for MongoDB.
- Used to connect, query, and manipulate MongoDB databases from Python applications.

-------------------------------------------------------------------------------------------------------------------------------------------------------
Q10: What are the ACID properties in the context of MongoDB transactions?
Ans:-
- Atomicity, Consistency, Isolation, Durability.
- MongoDB supports multi‑document transactions to ensure reliable operations.

-------------------------------------------------------------------------------------------------------------------------------------------------------
Q11: What is the purpose of MongoDB’s explain() function?
Ans:-
- Provides query execution details.
- Helps analyze performance and index usage.

-------------------------------------------------------------------------------------------------------------------------------------------------------
Q12: How does MongoDB handle schema validation?
Ans:-
- Allows defining rules for document structure.
- Ensures data integrity by enforcing required fields and types.

-------------------------------------------------------------------------------------------------------------------------------------------------------
Q13: What is the difference between a primary and a secondary node in a replica set?
Ans:-
- Primary: Accepts writes and reads.
- Secondary: Replicates data from primary, serves read queries if enabled.

-------------------------------------------------------------------------------------------------------------------------------------------------------
Q14: What security mechanisms does MongoDB provide for data protection?
Ans:-
- Authentication and authorization.
- Role‑based access control (RBAC).
- TLS/SSL encryption.
- Field‑level encryption.

-------------------------------------------------------------------------------------------------------------------------------------------------------
Q15: Explain the concept of embedded documents and when they should be used.
Ans:-
- Documents within documents (nested JSON).
- Useful when related data is frequently accessed together (e.g., user profile with address).

-------------------------------------------------------------------------------------------------------------------------------------------------------
Q16: What is the purpose of MongoDB’s $lookup stage in aggregation?
Ans:-
- Performs left outer joins between collections.
- Enables combining related data without SQL‑style joins.

-------------------------------------------------------------------------------------------------------------------------------------------------------
Q18: What are the advantages of using MongoDB for horizontal scaling?
Ans:-
- Sharding distributes data across servers.
- Enables handling large datasets and high throughput.
- Scales out easily as demand grows.

-------------------------------------------------------------------------------------------------------------------------------------------------------
Q19: How do MongoDB transactions differ from SQL transactions?
Ans:-
- SQL: Strict ACID by default.
- MongoDB: ACID supported via replica sets and transactions, but optimized for distributed workloads.

Q20: What are the main differences between capped collections and regular collections?
Ans:-
- Capped: Fixed size, auto‑overwrites oldest data, insertion order preserved.
- Regular: Flexible size, no overwrite behavior.

Q21: What is the purpose of the $match stage in MongoDB’s aggregation pipeline?
Ans:-
- Filters documents at the start of the pipeline.
- Improves efficiency by reducing data passed to later stages.

Q22: How can you secure access to a MongoDB database?
Ans:-
- Enable authentication.
- Use strong user roles.
- Restrict network access (IP whitelisting, firewalls).
- Encrypt data in transit and at rest.

Q23: What is MongoDB’s WiredTiger storage engine, and why is it important?
Ans:-
- Default storage engine in MongoDB.
- Provides document‑level concurrency, compression, and efficient memory usage.
- Critical for performance and scalability.
