Que.1)  What is Deep Learning? Briefly describe how it evolved and how it differs from traditional machine learning.
Ans:- Deep Learning is a subset of machine learning that uses artificial neural networks with multiple layers (hence "deep") to automatically learn hierarchical representations of data. It excels at handling complex, high-dimensional data such as images, speech, and text.

Evolution of Deep Learning:-
- 1950s–1980s: Early neural network models like the Perceptron and Backpropagation laid the foundation. However, limited computational power and small datasets restricted progress.
- 1990s–2000s: Neural networks lost popularity to traditional ML methods (e.g., SVMs, decision trees) due to better performance on small datasets.
- 2010s onward: With the rise of big data, GPUs, and improved algorithms, deep learning surged ahead. Breakthroughs in computer vision (CNNs), natural language processing (RNNs, Transformers), and speech recognition made it mainstream.

Que.2) Explain the basic architecture and functioning of a Perceptron. What are its limitations?
Ans:- A Perceptron is the simplest type of artificial neural network, introduced by Frank Rosenblatt in 1958. It models a single neuron and is mainly used for binary classification tasks.

Components:
1) Inputs (𝑥1,𝑥2,…,𝑥𝑛) – Features of the dataset.
2) Weights (𝑤1,𝑤2,…,𝑤𝑛) – Each input is multiplied by a weight that represents its importance.
3) Summation Function – Computes the weighted sum:𝑧=∑𝑖=1𝑛𝑤𝑖𝑥𝑖+𝑏
where 𝑏 is the bias term.
4) Activation Function – Applies a threshold (step function) to decide the output: 𝑦={1if 𝑧≥𝜃0if 𝑧<𝜃
This determines the class label.

Functioning of a Perceptron:-
Initialization: Assign random weights and bias.
Forward Pass: Compute weighted sum and apply activation function.
Prediction: Output is either 0 or 1 (binary classification).
Learning (Training): Adjust weights using the Perceptron Learning Rule:𝑤𝑖←𝑤𝑖+Δ𝑤𝑖

Limitations of a Perceptron:-
Linear Separability: Can only solve problems where data is linearly separable (e.g., AND, OR). It fails on non-linear problems like XOR.
Single Layer: Lacks hidden layers, so it cannot learn complex patterns.
Limited Expressiveness: Cannot capture relationships in high-dimensional or unstructured data (images, text, etc.).
Step Activation: The binary step function is non-differentiable, making gradient-based optimization impossible.

Que.3) Describe the purpose of activation function in neural networks. Compare Sigmoid, ReLU, and Tanh functions.
Ans:- Activation functions introduce non-linearity into neural networks. Without them, a network would just be a linear combination of inputs and weights, no matter how many layers it has. Non-linearity allows the network to learn complex patterns and approximate any function (thanks to the Universal Approximation Theorem).

Key Roles:
- Decide whether a neuron should be activated (fired).
- Enable the network to capture non-linear relationships.
- Control the flow of information through layers.
- Help in gradient-based optimization (backpropagation).

Que.4)  What is the difference between Loss function and Cost function in neural networks? Provide examples.
Ans:- Although often used interchangeably, loss function and cost function have distinct meanings in neural networks:

A) Loss Function
Definition: Measures the error for a single training example.
Purpose: Quantifies how far the predicted output is from the actual target for one data point.
Examples:
- Mean Squared Error (MSE) for regression
- Binary Cross-Entropy for classification

B) Cost Function
Definition: The average of the loss function over the entire training dataset.
Purpose: Provides a global measure of how well the model is performing across all samples.
Examples:
- Mean Squared Error (MSE) Cost
- Cross-Entropy Cost

Que.5) What is the role of optimizers in neural networks? Compare Gradient Descent, Adam, and RMSprop.  
Ans:- Optimizers are algorithms used to update the weights and biases of a neural network during training. Their role is to minimize the cost function by finding the optimal parameters. They determine how fast and in what direction the model learns.

Key Functions of Optimizers:-
- Adjust weights based on gradients (from backpropagation).
- Control learning speed (via learning rate).
- Help escape local minima and saddle points.
- Improve convergence efficiency and stability.

Comparison of Gradient Descent, RMSprop, and Adam:-
a) Gradient Descent is the most fundamental optimizer. It updates weights by moving them in the opposite direction of the gradient of the cost function. While conceptually simple, it can be slow and inefficient for large datasets or complex models. It also struggles with non-convex functions, where it may get stuck in local minima or saddle points.

b) RMSprop improves upon this by adapting the learning rate for each parameter. It keeps an exponentially decaying average of squared gradients, which helps stabilize training, especially in cases where gradients vary widely. This makes RMSprop particularly effective for recurrent neural networks and problems with noisy gradients. However, it requires careful tuning of hyperparameters and may converge to suboptimal solutions if not managed well.

c) Adam (Adaptive Moment Estimation) combines the strengths of both momentum and RMSprop. It calculates adaptive learning rates using estimates of the first moment (mean of gradients) and second moment (variance of gradients). This allows Adam to converge faster and more reliably than basic gradient descent. It has become the default choice in deep learning because it balances speed and robustness. Still, in some cases, Adam can overfit or fail to generalize as well as simpler methods like stochastic gradient descent.

```python
#Que. Write a Python program to implement a single-layer perceptron from scratch using NumPy to solve the logical AND gate.
#Ans:-
import numpy as np
import matplotlib.pyplot as plt

# AND gate dataset
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([0,0,0,1])

# Parameters
epochs = 20
lr = 0.1
weights = np.random.rand(2)   # two inputs
bias = np.random.rand(1)

# Activation function (step)
def step_function(z):
    return 1 if z >= 0 else 0

# Training loop
errors = []
for epoch in range(epochs):
    total_error = 0
    for i in range(len(X)):
        z = np.dot(X[i], weights) + bias
        y_pred = step_function(z)
        error = y[i] - y_pred
        # Update rule
        weights += lr * error * X[i]
        bias += lr * error
        total_error += abs(error)
    errors.append(total_error)
    print(f"Epoch {epoch+1}, Error: {total_error}")

# Final weights and bias
print("Weights:", weights)
print("Bias:", bias)

# Plot error trend
plt.plot(range(1, epochs+1), errors, marker='o')
plt.xlabel("Epochs")
plt.ylabel("Total Error")
plt.title("Error Trend for AND Gate Perceptron")
plt.show()
```

```python
#Que.7)  Implement and visualize Sigmoid, ReLU, and Tanh activation functions using Matplotlib.
#Ans:- 
import numpy as np
import matplotlib.pyplot as plt

# Define activation functions
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def relu(x):
    return np.maximum(0, x)

def tanh(x):
    return np.tanh(x)

# Input range
x = np.linspace(-10, 10, 100)

# Compute outputs
y_sigmoid = sigmoid(x)
y_relu = relu(x)
y_tanh = tanh(x)

# Plot all activation functions
plt.figure(figsize=(12, 8))

# Sigmoid
plt.subplot(3, 1, 1)
plt.plot(x, y_sigmoid, color='blue')
plt.title("Sigmoid Activation Function")
plt.grid(True)

# ReLU
plt.subplot(3, 1, 2)
plt.plot(x, y_relu, color='green')
plt.title("ReLU Activation Function")
plt.grid(True)

# Tanh
plt.subplot(3, 1, 3)
plt.plot(x, y_tanh, color='red')
plt.title("Tanh Activation Function")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
#Que.8) Use Keras to build and train a simple multilayer neural network on the MNIST digits dataset. Print the training accuracy. 
#Ans:-
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.utils import to_categorical

# 1. Load MNIST dataset
(X_train, y_train), (X_test, y_test) = mnist.load_data()

# 2. Normalize pixel values (0–255 → 0–1)
X_train = X_train / 255.0
X_test = X_test / 255.0

# 3. One-hot encode labels
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

# 4. Build a simple multilayer neural network
model = Sequential([
    Flatten(input_shape=(28, 28)),       # Flatten 28x28 images into 784 vector
    Dense(128, activation='relu'),       # Hidden layer with ReLU
    Dense(64, activation='relu'),        # Another hidden layer
    Dense(10, activation='softmax')      # Output layer (10 classes)
])

# 5. Compile the model
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# 6. Train the model
history = model.fit(X_train, y_train, epochs=5, batch_size=32, verbose=1)

# 7. Print training accuracy
train_acc = history.history['accuracy'][-1]
print(f"Final Training Accuracy: {train_acc*100:.2f}%")

# 8. Plot training accuracy over epochs
plt.plot(history.history['accuracy'], marker='o')
plt.title("Training Accuracy over Epochs")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.grid(True)
plt.show()
```

```python
#Que.9)  Visualize the loss and accuracy curves for a neural network model trained on the Fashion MNIST dataset. Interpret the training behavior.
#Ans:-
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.datasets import fashion_mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.utils import to_categorical

# 1. Load Fashion MNIST dataset
(X_train, y_train), (X_test, y_test) = fashion_mnist.load_data()

# 2. Normalize pixel values
X_train = X_train / 255.0
X_test = X_test / 255.0

# 3. One-hot encode labels
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

# 4. Build a simple neural network
model = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation='relu'),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
])

# 5. Compile the model
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# 6. Train the model
history = model.fit(X_train, y_train, epochs=10, batch_size=32,
                    validation_data=(X_test, y_test), verbose=1)

# 7. Plot loss and accuracy curves
plt.figure(figsize=(12, 5))

# Accuracy
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Training Accuracy', marker='o')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy', marker='x')
plt.title("Accuracy Curve")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)

# Loss
plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Training Loss', marker='o')
plt.plot(history.history['val_loss'], label='Validation Loss', marker='x')
plt.title("Loss Curve")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```
