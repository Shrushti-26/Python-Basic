#Que.1)  Compare and contrast NLTK and spaCy in terms of features, ease of use, and performance.
#Ans:-
A) Features
1. NLTK (Natural Language Toolkit)
- Rich set of linguistic resources: tokenizers, stemmers, lemmatizers, POS taggers, parsers.
- Includes corpora and lexical resources (WordNet).
- Great for educational/research use—you can experiment with many algorithms.
- More modular but requires manual setup of pipelines.

2. spaCy
- Industrial‑strength NLP library focused on production use.
- Pre‑trained statistical models for tokenization, POS tagging, dependency parsing, NER.
- Built‑in word vectors and integration with deep learning frameworks.
- Provides ready‑to‑use pipelines optimized for speed and accuracy.

B) Ease of Use
1. NLTK
- Steeper learning curve: requires piecing together multiple modules.
- More verbose code; good for learning fundamentals.
- Example: Tokenization often requires importing specific functions and resources separately.

2. spaCy
- Very user‑friendly: load a model once, then access tokens, entities, dependencies in a single pipeline.
- Minimal boilerplate code.
- Example: nlp = spacy.load("en_core_web_sm"); doc = nlp(text) gives you everything at once.

C) Performance
1. NLTK
- Slower, not optimized for large‑scale production.
- Better suited for small datasets, prototyping, or teaching.
- Many algorithms are rule‑based or classical ML, not modern deep learning.

2. spaCy
- Highly optimized in Cython for speed.
- Handles large volumes of text efficiently.
- Designed for real‑world applications (chatbots, information extraction, document processing).
- Supports GPU acceleration when integrated with deep learning frameworks.

#Que.4) Describe the architecture and functioning of a Recurrent Natural Network (RNN).
#Ans:-
A) Architecture of RNN
- Input Layer: Accepts sequential data (e.g., words in a sentence, time‑series values).
- Hidden Layer(s): Core of the RNN, where each hidden state ℎ𝑡depends on both the current input 𝑥𝑡 and the previous hidden state ℎ𝑡−1.
- Weights: Shared across time steps, unlike feedforward networks where each layer has separate weights.
- Output Layer: Produces predictions at each time step (e.g., next word, classification label).
- Mathematically:ℎ𝑡=𝑓(𝑊𝑥ℎ⋅𝑥𝑡+𝑊ℎℎ⋅ℎ𝑡−1+𝑏ℎ)𝑦𝑡=𝑔(𝑊ℎ𝑦⋅ℎ𝑡+𝑏𝑦)
Where:𝑥𝑡 = input at time step 𝑡ℎ𝑡 = hidden state at time step 𝑡𝑦𝑡 = output at time step 𝑡𝑓, 𝑔 = activation functions (commonly tanh, ReLU, or softmax)

B) Functioning of RNN
1. Sequential Processing
- RNNs process data step‑by‑step, maintaining a “memory” of previous inputs via hidden states.
- Example: In a sentence, the meaning of a word depends on the words before it.

2. Weight Sharing
- The same weights are applied across all time steps, making RNNs efficient for sequences of varying lengths.

3. Backpropagation Through Time (BPTT)
- Training involves unfolding the RNN across time steps and applying backpropagation.
- Gradients are propagated backward through the sequence to update weights.

4. Memory of Context
- Hidden states act as memory, capturing dependencies between past and current inputs.
- Useful for tasks like language modeling, speech recognition, and time‑series prediction.

#Que.5)  What is the key difference between LSTM and GRU networks in NLP applications?
#Ans:-
Key Difference:-
- LSTM uses three gates (input, forget, output) and a cell state to manage long‑term dependencies.
- GRU uses two gates (reset and update) and merges the hidden state and cell state into one, making it simpler and faster.

In NLP Applications:-
1. LSTM is preferred when:
- Long documents or sequences need modeling.
- Tasks require fine‑grained context retention (e.g., translation, speech recognition).

2. GRU is preferred when:
- Computational efficiency matters.
- Datasets are smaller or tasks are simpler (e.g., sentiment classification, intent detection).

```python
# Que.7)  Given the sample paragraph below, perform string tokenization and 
# frequency distribution using Python and NLTK: 
# “Natural Language Processing (NLP) is a fascinating field that combines linguistics, 
# computer science, and artificial intelligence. It enables machines to understand, 
# interpret, and generate human language. Applications of NLP include chatbots, 
# sentiment analysis, and machine translation. As technology advances, the role of NLP 
# in modern solutions is becoming increasingly critical.” 
#Ans:-
# Step 1: Install NLTK if not already installed
%pip install nltk

import nltk
from nltk.tokenize import word_tokenize
from nltk.probability import FreqDist

# Download required resources (only once)
nltk.download('punkt')

# Step 2: Define the sample paragraph
text = """Natural Language Processing (NLP) is a fascinating field that combines linguistics,
computer science, and artificial intelligence. It enables machines to understand,
interpret, and generate human language. Applications of NLP include chatbots,
sentiment analysis, and machine translation. As technology advances, the role of NLP
in modern solutions is becoming increasingly critical."""

# Step 3: Tokenization
tokens = word_tokenize(text)

print("Tokens:")
print(tokens)

# Step 4: Frequency Distribution
freq_dist = FreqDist(tokens)

print("\nFrequency Distribution (Top 10 words):")
for word, freq in freq_dist.most_common(10):
    print(f"{word}: {freq}")

# Step 5: Visualization
freq_dist.plot(20, cumulative=False, title="Word Frequency Distribution")
```

```python
# Que.8) Implement a basic LSTM model in Keras for a text classification task using 
# the following dummy dataset. Your model should classify sentences as either positive 
# (1) or negative (0). 
# # Dataset 
# texts = [ 
# “I love this project”, #Positive 
# “This is an amazing experience”, #Positive 
# “I hate waiting in line”, #Negative 
# “This is the worst service”, #Negative 
# “Absolutely fantastic!” #Positive 
# ] 
# labels = [1, 1, 0, 0, 1] 
# Preprocess the text, tokenize it, pad sequences, and build an LSTM model to train on 
# this data. You may use Keras with TensorFlow backend. 
#Ans:-
# Step 1: Install dependencies if needed
%pip install tensorflow

import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

# Step 2: Dummy dataset
texts = [
    "I love this project",              # Positive
    "This is an amazing experience",    # Positive
    "I hate waiting in line",           # Negative
    "This is the worst service",        # Negative
    "Absolutely fantastic!"             # Positive
]
labels = [1, 1, 0, 0, 1]

# Step 3: Tokenization
tokenizer = Tokenizer()
tokenizer.fit_on_texts(texts)
sequences = tokenizer.texts_to_sequences(texts)

# Step 4: Padding sequences
max_len = max(len(seq) for seq in sequences)
X = pad_sequences(sequences, maxlen=max_len, padding='post')
y = np.array(labels)

vocab_size = len(tokenizer.word_index) + 1

# Step 5: Build LSTM model
model = Sequential()
model.add(Embedding(input_dim=vocab_size, output_dim=16, input_length=max_len))
model.add(LSTM(32))
model.add(Dense(1, activation='sigmoid'))

# Step 6: Compile model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Step 7: Train model
model.fit(X, y, epochs=10, batch_size=2, verbose=1)

# Step 8: Evaluate / Predict
test_texts = ["I really enjoyed the service", "This was terrible"]
test_seq = tokenizer.texts_to_sequences(test_texts)
test_pad = pad_sequences(test_seq, maxlen=max_len, padding='post')

predictions = model.predict(test_pad)
print("Predictions:", predictions)
```

```python
# Que.9)  Using spaCy, build a simple NLP pipeline that includes tokenization, 
# lemmatization, and entity recognition. Use the following paragraph as your dataset: 
# “Homi Jehangir Bhaba was an Indian nuclear physicist who played a key role in the 
# development of India’s atomic energy program. He was the founding director of the Tata 
# Institute of Fundamental Research (TIFR) and was instrumental in establishing the 
# Atomic Energy Commission of India.” 
# Write a Python program that processes this text using spaCy, then prints tokens, their 
# lemmas, and any named entities found.
#Ans:-
# Step 1: Install spaCy if not already installed
%pip install spacy

import spacy

# Step 2: Load the English model
nlp = spacy.load("en_core_web_sm")

# Step 3: Define the text
text = """Homi Jehangir Bhaba was an Indian nuclear physicist who played a key role in the 
development of India’s atomic energy program. He was the founding director of the Tata 
Institute of Fundamental Research (TIFR) and was instrumental in establishing the 
Atomic Energy Commission of India."""

# Step 4: Process the text
doc = nlp(text)

# Step 5: Tokenization and Lemmatization
print("Tokens and Lemmas:")
for token in doc:
    print(f"{token.text} → {token.lemma_}")

# Step 6: Named Entity Recognition
print("\nNamed Entities:")
for ent in doc.ents:
    print(f"{ent.text} ({ent.label_})")
```

```python
# Que.10) You are working on a chatbot for a mental health platform. Explain how 
# you would leverage LSTM or GRU networks along with libraries like spaCy or Stanford 
# NLP to understand and respond to user input effectively. Detail your architecture, data 
# preprocessing pipeline, and any ethical considerations. 
#Ans:-
# Step 1: Install dependencies if needed
# %pip install spacy tensorflow

import spacy
import numpy as np
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, GRU, Dense

# Step 2: Load spaCy model for preprocessing
nlp = spacy.load("en_core_web_sm")

# Example dataset (dummy intents: stress, positive, neutral)
texts = [
    "I feel anxious about exams",       # stress
    "I am happy with my progress",      # positive
    "I am just tired today",            # neutral
]
labels = [0, 1, 2]  # 0=stress, 1=positive, 2=neutral

# Step 3: Preprocess with spaCy (lemmatization)
def preprocess(text):
    doc = nlp(text.lower())
    return " ".join([token.lemma_ for token in doc if not token.is_stop])

processed_texts = [preprocess(t) for t in texts]

# Step 4: Tokenization + Padding
tokenizer = Tokenizer()
tokenizer.fit_on_texts(processed_texts)
sequences = tokenizer.texts_to_sequences(processed_texts)

max_len = max(len(seq) for seq in sequences)
X = pad_sequences(sequences, maxlen=max_len, padding='post')
y = np.array(labels)

vocab_size = len(tokenizer.word_index) + 1

# Step 5: Build GRU model
model = Sequential()
model.add(Embedding(input_dim=vocab_size, output_dim=16, input_length=max_len))
model.add(GRU(32))
model.add(Dense(3, activation='softmax'))  # 3 classes

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

# Step 6: Train model
model.fit(X, y, epochs=10, batch_size=2, verbose=1)

# Step 7: Test prediction
test_text = "I feel overwhelmed with work"
test_proc = preprocess(test_text)
test_seq = tokenizer.texts_to_sequences([test_proc])
test_pad = pad_sequences(test_seq, maxlen=max_len, padding='post')

pred = model.predict(test_pad)
print("Prediction probabilities:", pred)
print("Predicted class:", np.argmax(pred))
```
