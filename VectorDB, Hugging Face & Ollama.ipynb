#Que.1) What is a Vector Database (VectorDB) and how is it different from traditional databases?
#Ans:- A Vector Database (VectorDB) is a specialized type of database designed to store, index, and query vector embeddings—numerical representations of data such as text, images, audio, or video. These embeddings are typically generated by machine learning models (e.g., Hugging Face transformers, OpenAI embeddings, etc) and capture semantic meaning in high-dimensional space.

Key Features of Vector Databases:-
- Stores embeddings (vectors): Instead of rows of structured data, it stores high-dimensional vectors (e.g., 512–4096 dimensions).
- Similarity search: Enables fast queries like “find the most similar documents/images to this input” using metrics such as cosine similarity, Euclidean distance, or dot product.
- Optimized indexing: Uses advanced data structures (e.g., HNSW graphs, IVF, PQ) for efficient nearest neighbor search.
- Scales for AI workloads: Designed for billions of vectors, supporting real-time retrieval in applications like recommendation systems, semantic search, and generative AI.

#Que.2) Explain the various types of VectorDBs available and describe their suitability for different use cases. 
#Ans:- 
Types of Vector Databases:-
A) Embedded / Lightweight Vector Stores:-
- Examples: FAISS (Facebook AI Similarity Search), Annoy (Spotify), ScaNN (Google)
- Characteristics:
  - Library-based, runs inside your application.
  - Optimized for speed and memory efficiency.
  - No distributed infrastructure—best for local or small-scale projects.
- Use Cases:
  - Academic research, prototypes, Jupyter notebooks.
  - Personal projects (semantic search, clustering).
  - Offline or single-machine workloads.

B) Cloud-Native Vector Databases:-
- Examples: Pinecone, Weaviate, Milvus, Qdrant
- Characteristics:
  - Managed services with APIs.
  - Scalable to billions of vectors.
  - Support metadata filtering + hybrid search (combine keyword + vector search).
- Use Cases:
  - Production-grade semantic search engines.
  - AI-powered recommendation systems.
  - Enterprise-scale retrieval-augmented generation (RAG) pipelines.

C) Hybrid Databases (Traditional DBs with Vector Support):-
- Examples: PostgreSQL + pgvector, MongoDB Atlas Vector Search, ElasticSearch with vector fields
- Characteristics:
  - Extend existing relational/NoSQL DBs with vector indexing.
  - Allow mixing structured queries (WHERE age > 30) with similarity search.
  - Lower learning curve if you already use these DBs.
- Use Cases:
  - Applications needing both structured queries and semantic search.
  - Enterprise systems where vector search is an add-on, not the core.
  - Customer data + embeddings in one place.

D) Specialized Vector Engines:-
- Examples: Vespa, Redis Vector Search
- Characteristics:
  - Built for real-time, low-latency retrieval.
  - Often integrated with caching or streaming systems.
  - Focus on speed and operational simplicity.
- Use Cases:
  - Real-time personalization (ads, e-commerce).
  - Chatbots needing instant context retrieval.
  - High-throughput recommendation pipelines.

```python
#Que.7)  Demonstrate how to download and fine-tune a Hugging Face model for a text generation task. 
#Ans:-
# Step 1: Install dependencies
%pip install --upgrade transformers datasets accelerate

# Step 2: Load model + tokenizer
from transformers import AutoTokenizer, AutoModelForCausalLM

model_name = "gpt2"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

# Fix padding issue (GPT-2 has no pad token)
tokenizer.pad_token = tokenizer.eos_token
model.resize_token_embeddings(len(tokenizer))

# Step 3: Prepare dataset
from datasets import Dataset

texts = [
    {"text": "AI is transforming industries."},
    {"text": "Machine learning enables predictive analytics."},
    {"text": "Generative AI creates new possibilities."}
]

dataset = Dataset.from_list(texts)

def tokenize_function(examples):
    encodings = tokenizer(
        examples["text"],
        truncation=True,
        padding="max_length",
        max_length=64
    )
    # Add labels for causal LM training
    encodings["labels"] = encodings["input_ids"].copy()
    return encodings

tokenized_dataset = dataset.map(tokenize_function, batched=True)

# Step 4: Define training arguments
from transformers import TrainingArguments, Trainer

training_args = TrainingArguments(
    output_dir="./results",
    learning_rate=2e-5,
    per_device_train_batch_size=2,
    num_train_epochs=2,
    weight_decay=0.01,
    save_strategy="epoch",
    logging_dir="./logs",
    logging_steps=10
)

# Step 5: Fine-tune with Trainer
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_dataset
)

trainer.train()

# Step 6: Generate text with fine-tuned model
from transformers import pipeline

generator = pipeline("text-generation", model=model, tokenizer=tokenizer)

output = generator("Generative AI will", max_length=50, num_return_sequences=1)
print(output[0]['generated_text'])
```

```python
#Que.9) : Implement a basic RAG (Retrieval-Augmented Generation) system using Ollama with Llama3.
#Ans:-
# Step 1: Install dependencies
%pip install chromadb sentence-transformers ollama

# Step 2: Import libraries
import chromadb
from sentence_transformers import SentenceTransformer
import ollama

# Step 3: Initialize ChromaDB (persistent storage)
client = chromadb.PersistentClient(path="chroma_db")
collection = client.get_or_create_collection("rag_collection")

# Step 4: Load embedding model (Hugging Face)
embedder = SentenceTransformer("all-MiniLM-L6-v2")

# Step 5: Insert sample documents into ChromaDB
documents = [
    "AI is transforming industries by automating tasks.",
    "Machine learning enables predictive analytics in business.",
    "Vector databases store embeddings for semantic search.",
    "Ollama allows running LLMs like Llama3 locally."
]

embeddings = embedder.encode(documents).tolist()

collection.add(
    ids=[f"doc{i}" for i in range(len(documents))],
    documents=documents,
    embeddings=embeddings
)

# Step 6: Define RAG query function
def rag_query(user_query):
    # Generate embedding for query
    query_embedding = embedder.encode([user_query]).tolist()

    # Retrieve top 2 relevant docs
    results = collection.query(
        query_embeddings=query_embedding,
        n_results=2
    )

    context = " ".join(results["documents"][0])

    # Pass context + query to Ollama (Llama3)
    prompt = f"Context: {context}\n\nQuestion: {user_query}\n\nAnswer:"
    response = ollama.chat(model="llama3", messages=[
        {"role": "user", "content": prompt}
    ])

    return response["message"]["content"]

# Step 7: Run a sample query
answer = rag_query("Explain how vector databases help in RAG systems.")
print("RAG Answer:\n", answer)
```

```python
# Que.10)  A health-tech startup wants to build a chatbot that can answer user 
# queries based on medical research articles. Propose and explain a solution using 
# Hugging Face models for understanding, VectorDB for retrieval, and Ollama for 
# generation.
#Ans:-
# Step 1: Install dependencies
#%pip install chromadb sentence-transformers ollama

# Step 2: Imports
import chromadb
from sentence_transformers import SentenceTransformer
import ollama

# Step 3: Initialize ChromaDB
client = chromadb.PersistentClient(path="medical_db")
collection = client.get_or_create_collection("medical_articles")

# Step 4: Load embedding model
embedder = SentenceTransformer("all-MiniLM-L6-v2")

# Step 5: Insert sample medical research chunks
documents = [
    "Recent studies show AI can assist in early cancer detection.",
    "Machine learning models improve accuracy in diagnosing heart disease.",
    "Vector databases enable semantic search across medical literature."
]

embeddings = embedder.encode(documents).tolist()

collection.add(
    ids=[f"doc{i}" for i in range(len(documents))],
    documents=documents,
    embeddings=embeddings
)

# Step 6: Define RAG query function
def rag_query(user_query):
    # Embed query
    query_embedding = embedder.encode([user_query]).tolist()
    
    # Retrieve relevant docs
    results = collection.query(
        query_embeddings=query_embedding,
        n_results=2
    )
    
    context = " ".join(results["documents"][0])
    
    # Build prompt for Llama3
    prompt = f"Context: {context}\n\nQuestion: {user_query}\n\nAnswer:"
    
    response = ollama.chat(model="llama3", messages=[
        {"role": "user", "content": prompt}
    ])
    
    return response["message"]["content"]

# Step 7: Run a sample query
answer = rag_query("How is AI used in cancer detection?")
print("Chatbot Answer:\n", answer)
```
